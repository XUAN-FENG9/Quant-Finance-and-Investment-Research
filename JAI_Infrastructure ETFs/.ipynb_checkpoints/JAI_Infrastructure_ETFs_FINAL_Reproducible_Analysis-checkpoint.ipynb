```python
CUT = pd.Timestamp("2026-04-30")

crsp["date"] = pd.to_datetime(crsp["caldt"]) + pd.offsets.MonthEnd(0)
for c in ["mret","mtna","mnav"]:
    crsp[c] = pd.to_numeric(crsp[c], errors="coerce")

univ["crsp_fundno"] = univ["crsp_fundno"].astype(int)
univ["first_offer_dt"] = pd.to_datetime(univ["first_offer_dt"], errors="coerce")
univ["earliest_theme_snapshot"] = pd.to_datetime(univ["earliest_theme_snapshot"], errors="coerce")

def final_category(row):
    pc, tk = row["provisional_category"], str(row["ticker"])
    if tk in ("IFRA","GREI","BLLD"): return "Hybrid"
    if tk == "BILT": return "Owner/operator"
    if tk == "ENRG": return "Renewable/power/grid"
    mapping = {
        "Broad owner/operator infrastructure":"Owner/operator",
        "Development/enabler infrastructure":"Development/enabler",
        "Digital/data infrastructure":"Digital/data",
        "Energy infrastructure":"Energy networks",
        "Renewable/power/grid infrastructure":"Renewable/power/grid",
        "Water infrastructure/adjacent":"Water",
        "Hybrid real-asset infrastructure":"Hybrid",
    }
    return mapping.get(pc,"Other")

univ["category"] = univ.apply(final_category,axis=1)

exact_theme_dates = {"RBLD":"2022-08-01","IDGT":"2023-12-18","POWR":"2025-10-29"}
def get_theme_start(row):
    tk = str(row["ticker"])
    if tk in exact_theme_dates:
        return pd.Timestamp(exact_theme_dates[tk]) + pd.offsets.MonthEnd(0)
    if tk == "ZAP":
        return row["first_offer_dt"] + pd.offsets.MonthEnd(0)
    if row["strategy_change_flag"] == "Yes":
        return row["earliest_theme_snapshot"] + pd.offsets.MonthEnd(0)
    return row["first_offer_dt"] + pd.offsets.MonthEnd(0)

univ["theme_start"] = univ.apply(get_theme_start,axis=1)

panel = crsp.merge(
    univ[["crsp_fundno","ticker","fund_name_latest","tier","category","dead_flag",
          "delist_cd","theme_start"]],
    on="crsp_fundno",how="inner"
)
panel = panel[(panel["date"]<=CUT)&(panel["date"]>=panel["theme_start"])&(panel["category"]!="Other")].copy()

def apply_tna_entry(panel, tna_min=5):
    pieces=[]
    for fid,g in panel.groupby("crsp_fundno"):
        g=g.sort_values("date")
        hit=g[(g["mtna"]>=tna_min)&g["mret"].notna()]
        if len(hit):
            pieces.append(g[g["date"]>=hit["date"].iloc[0]])
    return pd.concat(pieces,ignore_index=True)

eligible = apply_tna_entry(panel,5)
counts = eligible.groupby("crsp_fundno")["mret"].count()
risk_ids = counts[counts>=24].index
risk_panel = eligible[eligible["crsp_fundno"].isin(risk_ids)].copy()

print("Candidates:", len(univ))
print("Reached $5m:", eligible["crsp_fundno"].nunique())
print("24-month primary funds:", len(risk_ids))
print("Inactive/delisted in 24m sample:", risk_panel.drop_duplicates("crsp_fundno")["dead_flag"].eq("Y").sum())
```

## 4. Prepare Fama-French, 10-year yield, and VIX monthly data

```python
ff = ff5.copy()
ff["date"] = pd.to_datetime(ff["Date"].astype(str), format="%Y%m") + pd.offsets.MonthEnd(0)
for c in ["Mkt-RF","SMB","HML","RMW","CMA","RF"]:
    ff[c] = pd.to_numeric(ff[c],errors="coerce")/100

mm = mom.copy()
mm["date"] = pd.to_datetime(mm["Date"].astype(str), format="%Y%m") + pd.offsets.MonthEnd(0)
mm["MOM"] = pd.to_numeric(mm["MOM"],errors="coerce")/100
ff = ff.merge(mm[["date","MOM"]],on="date",how="inner")

yy=y10.copy()
yy["obs_date"]=pd.to_datetime(yy["observation_date"])
yy["THREEFY10"]=pd.to_numeric(yy["THREEFY10"],errors="coerce")
ym=yy.set_index("obs_date")["THREEFY10"].resample("ME").last().to_frame("y10").reset_index().rename(columns={"obs_date":"date"})
ym["d10y"]=ym["y10"].diff()

vv=vix.copy()
vv["obs_date"]=pd.to_datetime(vv["date"])
vv["VIX"]=pd.to_numeric(vv["VIX"],errors="coerce")
vm=vv.set_index("obs_date")["VIX"].resample("ME").agg(["last","mean","max"]).reset_index().rename(
    columns={"obs_date":"date","last":"vix_end","mean":"vix_mean","max":"vix_max"})

factors=ff.merge(ym[["date","y10","d10y"]],on="date",how="left").merge(
    vm[["date","vix_end","vix_mean","vix_max"]],on="date",how="left")
factors=factors[factors["date"]<=CUT].copy()
print("Factor end date:", factors["date"].max().date())
```

## 5. Fund-level risk, return, downside, beta, and income proxy

```python
risk_panel=risk_panel.sort_values(["crsp_fundno","date"]).copy()
risk_panel["prev_nav"]=risk_panel.groupby("crsp_fundno")["mnav"].shift(1)
risk_panel["prev_date"]=risk_panel.groupby("crsp_fundno")["date"].shift(1)
risk_panel["nav_ret"]=risk_panel["mnav"]/risk_panel["prev_nav"]-1
risk_panel["month_gap"]=(risk_panel["date"].dt.to_period("M")-risk_panel["prev_date"].dt.to_period("M")).apply(
    lambda x:x.n if pd.notna(x) else np.nan)
risk_panel["dist_proxy"]=(1+risk_panel["mret"])/(1+risk_panel["nav_ret"])-1
risk_panel.loc[(risk_panel["month_gap"]!=1)|(risk_panel["dist_proxy"]<-.02)|(risk_panel["dist_proxy"]>.10),"dist_proxy"]=np.nan
risk_panel["dist_proxy_pos"]=risk_panel["dist_proxy"].clip(lower=0)

def series_stats(r,rf):
    z=pd.concat([r.rename("r"),rf.rename("rf")],axis=1).dropna()
    r,rf=z["r"],z["rf"]; n=len(r)
    ann=np.prod(1+r)**(12/n)-1
    vol=r.std(ddof=1)*np.sqrt(12)
    rf_ann=np.prod(1+rf)**(12/n)-1
    sharpe=(ann-rf_ann)/vol
    downside=np.sqrt(np.mean(np.minimum(r,0)**2)*12)
    sortino=(ann-rf_ann)/downside
    wealth=(1+r).cumprod()
    mdd=(wealth/wealth.cummax()-1).min()
    return ann,vol,sharpe,sortino,mdd,downside

rows=[]
for fid,g in risk_panel.groupby("crsp_fundno"):
    z=g[["date","mret"]].merge(factors[["date","RF","Mkt-RF"]],on="date",how="inner").dropna()
    st=series_stats(z.set_index("date")["mret"],z.set_index("date")["RF"])
    ex=z["mret"]-z["RF"]
    beta=np.cov(ex,z["Mkt-RF"],ddof=1)[0,1]/np.var(z["Mkt-RF"],ddof=1)
    info=g.iloc[-1]
    rows.append([fid,info["ticker"],info["category"],info["tier"],info["dead_flag"],len(z),*st,beta,g["dist_proxy_pos"].mean()*12])

fund_metrics=pd.DataFrame(rows,columns=[
    "crsp_fundno","ticker","category","tier","dead_flag","months",
    "ann_return","ann_vol","sharpe","sortino","max_drawdown","downside_dev",
    "market_beta","income_proxy"
])

category_metrics=fund_metrics.groupby("category").agg(
    n=("ticker","size"),dead=("dead_flag",lambda s:(s=="Y").sum()),
    ann_return=("ann_return","median"),ann_vol=("ann_vol","median"),
    sharpe=("sharpe","median"),sortino=("sortino","median"),
    max_drawdown=("max_drawdown","median"),market_beta=("market_beta","median"),
    income_proxy=("income_proxy","median")).reset_index()

display(category_metrics.round(4))

for metric in ["ann_return","ann_vol","sharpe","sortino","max_drawdown","market_beta","income_proxy"]:
    groups=[g[metric].dropna().values for _,g in fund_metrics.groupby("category")]
    print(metric, "Kruskal p =", round(stats.kruskal(*groups).pvalue,6))

# Holm-adjusted pairwise Mann-Whitney follow-ups for metrics with significant omnibus tests.
from itertools import combinations as _pair_combinations

def holm_adjust(pvals):
    pvals = np.asarray(pvals, dtype=float)
    order = np.argsort(pvals)
    adjusted = np.empty_like(pvals)
    running = 0.0
    m = len(pvals)
    for rank, idx in enumerate(order):
        val = (m-rank) * pvals[idx]
        running = max(running, val)
        adjusted[idx] = min(running, 1.0)
    return adjusted

pairwise_rows = []
for metric in ["ann_vol","market_beta","income_proxy"]:
    tmp = []
    cats_pw = sorted(fund_metrics["category"].unique())
    for a,b in _pair_combinations(cats_pw,2):
        x = fund_metrics.loc[fund_metrics.category==a, metric].dropna()
        y = fund_metrics.loc[fund_metrics.category==b, metric].dropna()
        u,pv = stats.mannwhitneyu(x,y,alternative="two-sided")
        tmp.append([metric,a,b,u,pv])
    adj = holm_adjust([x[4] for x in tmp])
    for row,pa in zip(tmp,adj):
        pairwise_rows.append(row+[pa])

pairwise_holm = pd.DataFrame(
    pairwise_rows,
    columns=["metric","category_1","category_2","U","p_raw","p_holm"]
)
display(pairwise_holm.sort_values(["metric","p_holm"]).round(4))
```

## 6. Within-group homogeneity, between-group separation, and category return portfolios

```python
piv=risk_panel.pivot_table(index="date",columns="crsp_fundno",values="mret")
catdict=fund_metrics.set_index("crsp_fundno")["category"].to_dict()
pair_rows=[]
ids=list(fund_metrics["crsp_fundno"])
for i,j in combinations(ids,2):
    z=piv[[i,j]].dropna()
    if len(z)>=24:
        pair_rows.append([i,j,catdict[i],catdict[j],len(z),z[i].corr(z[j]),catdict[i]==catdict[j]])
corr_pairs=pd.DataFrame(pair_rows,columns=["fund1","fund2","cat1","cat2","overlap","corr","within"])
within=corr_pairs.loc[corr_pairs.within,"corr"]
between=corr_pairs.loc[~corr_pairs.within,"corr"]
print("Within median:",within.median(),"Between median:",between.median())
print("One-sided Mann-Whitney p:",stats.mannwhitneyu(within,between,alternative="greater").pvalue)

# No ex-post 24m requirement for category portfolios: use all contemporaneously eligible funds.
cat_returns={}
for cat in sorted(eligible["category"].unique()):
    cat_returns[cat]=eligible[eligible.category==cat].groupby("date")["mret"].mean()
category_returns=pd.DataFrame(cat_returns).sort_index()
common_returns=category_returns.dropna()
print("Common period:",common_returns.index.min().date(),"to",common_returns.index.max().date(),
      "| months:",len(common_returns))
display(common_returns.corr().round(2))
```

## 7. Multi-factor regressions and return-derived cluster validation

```python
fac_cols=["Mkt-RF","SMB","HML","RMW","CMA","MOM","d10y"]
factor_ids=fund_metrics.loc[fund_metrics.months>=36,"crsp_fundno"].tolist()
rows=[]
for fid in factor_ids:
    g=risk_panel[risk_panel.crsp_fundno==fid][["date","mret","ticker","category"]]
    z=g.merge(factors[["date","RF"]+fac_cols],on="date",how="inner").dropna()
    if len(z)<36: continue
    y=z["mret"]-z["RF"]
    X=sm.add_constant(z[fac_cols])
    res=sm.OLS(y,X).fit(cov_type="HAC",cov_kwds={"maxlags":3})
    row={"crsp_fundno":fid,"ticker":z.ticker.iloc[0],"category":z.category.iloc[0],
         "nobs":len(z),"alpha_ann":res.params["const"]*12,"alpha_p":res.pvalues["const"],
         "adj_r2":res.rsquared_adj}
    for c in fac_cols:
        row[c]=res.params[c]; row[c+"_p"]=res.pvalues[c]
    rows.append(row)
factor_regs=pd.DataFrame(rows)

summary=factor_regs.groupby("category").agg(
    n=("ticker","size"),alpha_ann=("alpha_ann","median"),adj_r2=("adj_r2","median"),
    MKT=("Mkt-RF","median"),SMB=("SMB","median"),HML=("HML","median"),
    RMW=("RMW","median"),CMA=("CMA","median"),MOM=("MOM","median"),Rate10Y=("d10y","median"))
display(summary.round(4))

for metric in ["alpha_ann","Mkt-RF","SMB","HML","RMW","CMA","MOM","d10y"]:
    groups=[g[metric].dropna().values for _,g in factor_regs.groupby("category")]
    print(metric,"Kruskal p =",round(stats.kruskal(*groups).pvalue,6))

Z=(factor_regs[fac_cols]-factor_regs[fac_cols].mean())/factor_regs[fac_cols].std(ddof=0)
clusters=AgglomerativeClustering(n_clusters=7,linkage="ward").fit_predict(Z)
true_codes=pd.Categorical(factor_regs["category"]).codes
ari=adjusted_rand_score(true_codes,clusters)
sil=silhouette_score(Z,clusters)
rng=np.random.default_rng(42)
perm=np.array([adjusted_rand_score(rng.permutation(true_codes),clusters) for _ in range(2000)])
p_perm=(1+(perm>=ari).sum())/(len(perm)+1)
print("ARI:",ari,"Silhouette:",sil,"Permutation p:",p_perm)
```

## 8. Downside and regime analysis

```python
fac_idx=factors.set_index("date").reindex(common_returns.index)
mkt_total=fac_idx["Mkt-RF"]+fac_idx["RF"]
bottom=mkt_total.quantile(.10)
rows=[]
for cat in common_returns.columns:
    s=common_returns[cat]
    covid=(1+s.loc[(s.index>="2020-02-29")&(s.index<="2020-04-30")]).prod()-1
    tight=(1+s.loc[(s.index>="2022-03-31")&(s.index<="2022-12-31")]).prod()-1
    rows.append([cat,covid,tight,s[fac_idx["vix_mean"]>=30].mean(),
                 s[fac_idx["d10y"]>=.25].mean(),s[mkt_total<=bottom].mean()])
regimes=pd.DataFrame(rows,columns=[
    "category","COVID_Feb_Apr_2020","2022_tightening_Mar_Dec",
    "high_VIX_month","rate_shock_month","bottom_decile_equity_month"])
display(regimes.round(4))
```

## 9. Portfolio risk contribution: fixed 60/30/10 case

```python
fall=factors.set_index("date")
fac_idx=fall.reindex(common_returns.index).copy()
fac_idx["US_equity"]=fac_idx["Mkt-RF"]+fac_idx["RF"]
# Stylized 10-year Treasury proxy: lagged yield carry minus duration (7) times yield change.
fac_idx["bond_proxy"]=(fall["y10"].shift(1)/100/12 - 7.0*(fall["d10y"]/100)).reindex(common_returns.index)
portfolio_data=pd.concat([fac_idx[["US_equity","bond_proxy","RF"]],common_returns],axis=1).dropna()

base=.6*portfolio_data["US_equity"]+.4*portfolio_data["bond_proxy"]
print("60/40 stats:",series_stats(base,portfolio_data["RF"])[:5])

rows=[]
for cat in common_returns.columns:
    z=portfolio_data[["US_equity","bond_proxy","RF",cat]]
    p=.6*z["US_equity"]+.3*z["bond_proxy"]+.1*z[cat]
    st=series_stats(p,z["RF"])
    cov=z[["US_equity","bond_proxy",cat]].cov().values
    w=np.array([.6,.3,.1])
    pvol=np.sqrt(w@cov@w)
    mrc=cov@w/pvol
    component=w*mrc
    rows.append([cat,*st[:5],z[cat].corr(.6*z["US_equity"]+.3*z["bond_proxy"]),
                 mrc[2]*np.sqrt(12),component[2]/pvol])
portfolio_case=pd.DataFrame(rows,columns=[
    "category","return","vol","sharpe","sortino","max_drawdown",
    "corr_core","marginal_risk_ann","component_risk_pct"])
display(portfolio_case.round(4))
```

## 10. Objective-dependent 10% infrastructure sleeve (constrained MVO illustration)

```python
cats=list(common_returns.columns)
R=portfolio_data[["US_equity","bond_proxy"]+cats].values
lw=LedoitWolf().fit(R)
Sigma=lw.covariance_; mu=R.mean(axis=0); rf_m=portfolio_data["RF"].mean()

def full_weights(v): return np.r_[.6,.3,.1*np.array(v)]
def pvar(v):
    w=full_weights(v); return w@Sigma@w
def pmu(v): return full_weights(v)@mu

bounds=[(0,.6)]*len(cats)
cons=[{"type":"eq","fun":lambda v:np.sum(v)-1}]
x0=np.ones(len(cats))/len(cats)

SOLVER_OPTIONS={"ftol":1e-12,"maxiter":500,"disp":False}

growth=optimize.minimize(lambda v:-(pmu(v)-rf_m)/np.sqrt(pvar(v)),x0,bounds=bounds,constraints=cons,method="SLSQP",options=SOLVER_OPTIONS)
defensive=optimize.minimize(lambda v:np.sqrt(pvar(v)),x0,bounds=bounds,constraints=cons,method="SLSQP",options=SOLVER_OPTIONS)

income_vec=category_metrics.set_index("category").reindex(cats)["income_proxy"].values
minvol_ann=np.sqrt(pvar(defensive.x))*np.sqrt(12)
income_cap=minvol_ann+.005
income_cons=cons+[{"type":"ineq","fun":lambda v:income_cap/np.sqrt(12)-np.sqrt(pvar(v))}]
income=optimize.minimize(lambda v:-v@income_vec,x0,bounds=bounds,constraints=income_cons,method="SLSQP",options=SOLVER_OPTIONS)

C=portfolio_data[cats].cov().values
core=.6*portfolio_data["US_equity"]+.3*portfolio_data["bond_proxy"]
cov_ci=np.array([portfolio_data[c].cov(core) for c in cats])
var_core=core.var()
def corr_obj(v): return (v@cov_ci)/np.sqrt((v@C@v)*var_core)
diversification=optimize.minimize(corr_obj,x0,bounds=bounds,constraints=cons,method="SLSQP",options=SOLVER_OPTIONS)

for _name,_res in [("Growth",growth),("Defensive",defensive),("Income",income),("Diversification",diversification)]:
    assert _res.success, f"{_name} optimization failed: {_res.message}"

solutions={
    "Growth":growth.x,"Defensive":defensive.x,"Income":income.x,"Diversification":diversification.x
}
weights=pd.DataFrame(solutions,index=cats).T
display(weights.round(4))

for obj,v in solutions.items():
    sleeve=sum(v[k]*portfolio_data[c] for k,c in enumerate(cats))
    p=.6*portfolio_data["US_equity"]+.3*portfolio_data["bond_proxy"]+.1*sleeve
    print(obj,series_stats(p,portfolio_data["RF"])[:5],"sleeve-core corr",sleeve.corr(core))
```

## 11. Robustness checks

```python
# Full robustness analysis reported in Exhibit 9.

def build_robustness_sample(tna_min=5, tier_a_only=False, live_only=False):
    p = panel.copy()
    if tier_a_only:
        p = p[p["tier"]=="A_core_explicit_infrastructure"]
    if live_only:
        p = p[p["dead_flag"]!="Y"]

    e = apply_tna_entry(p, tna_min)
    ids = e.groupby("crsp_fundno")["mret"].count()
    ids = ids[ids>=24].index
    rp = e[e["crsp_fundno"].isin(ids)].copy().sort_values(["crsp_fundno","date"])

    rp["prev_nav"] = rp.groupby("crsp_fundno")["mnav"].shift(1)
    rp["prev_date"] = rp.groupby("crsp_fundno")["date"].shift(1)
    rp["nav_ret"] = rp["mnav"]/rp["prev_nav"] - 1
    rp["month_gap"] = (
        rp["date"].dt.to_period("M") - rp["prev_date"].dt.to_period("M")
    ).apply(lambda x: x.n if pd.notna(x) else np.nan)
    rp["dist_proxy"] = (1+rp["mret"])/(1+rp["nav_ret"]) - 1
    rp.loc[
        (rp["month_gap"]!=1) | (rp["dist_proxy"]<-.02) | (rp["dist_proxy"]>.10),
        "dist_proxy"
    ] = np.nan
    rp["dist_proxy_pos"] = rp["dist_proxy"].clip(lower=0)

    rows = []
    for fid,g in rp.groupby("crsp_fundno"):
        z = g[["date","mret"]].merge(
            factors[["date","RF","Mkt-RF"]], on="date", how="inner"
        ).dropna()
        if len(z)<24:
            continue
        st = series_stats(z.set_index("date")["mret"], z.set_index("date")["RF"])
        ex = z["mret"] - z["RF"]
        beta = np.cov(ex,z["Mkt-RF"],ddof=1)[0,1] / np.var(z["Mkt-RF"],ddof=1)
        rows.append([
            fid,g["category"].iloc[-1],g["dead_flag"].iloc[-1],
            st[1],beta,g["dist_proxy_pos"].mean()*12
        ])

    fm = pd.DataFrame(rows,columns=["fid","category","dead","ann_vol","beta","income"])
    pvals = {}
    for met in ["ann_vol","beta","income"]:
        groups = [gg[met].dropna().values for _,gg in fm.groupby("category")]
        pvals[met] = stats.kruskal(*groups).pvalue

    owner = fm[fm.category=="Owner/operator"]
    dev = fm[fm.category=="Development/enabler"]
    return {
        "n_funds":len(fm),
        "n_dead":int((fm.dead=="Y").sum()),
        "p_vol":pvals["ann_vol"],
        "p_beta":pvals["beta"],
        "p_income":pvals["income"],
        "owner_median_vol":owner.ann_vol.median(),
        "owner_median_beta":owner.beta.median(),
        "dev_median_vol":dev.ann_vol.median(),
        "dev_median_beta":dev.beta.median(),
    }

robustness_rows = []
for label,kwargs in [
    ("Primary: TNA >= $5m",dict(tna_min=5,tier_a_only=False,live_only=False)),
    ("TNA >= $10m",dict(tna_min=10,tier_a_only=False,live_only=False)),
    ("Tier A explicit infrastructure only",dict(tna_min=5,tier_a_only=True,live_only=False)),
    ("Live funds only",dict(tna_min=5,tier_a_only=False,live_only=True)),
]:
    row = build_robustness_sample(**kwargs)
    row["specification"] = label
    robustness_rows.append(row)

robustness = pd.DataFrame(robustness_rows)[[
    "specification","n_funds","n_dead","p_vol","p_beta","p_income",
    "owner_median_vol","owner_median_beta","dev_median_vol","dev_median_beta"
]]
display(robustness.round(4))

# Equal-weight versus AUM-weighted category portfolios.
aum_ret = {}
for cat in sorted(eligible["category"].unique()):
    gg = eligible[eligible.category==cat].copy()
    def wtret(x):
        x = x.dropna(subset=["mret","mtna"])
        x = x[x.mtna>0]
        return np.average(x.mret,weights=x.mtna) if len(x) else np.nan
    aum_ret[cat] = gg.groupby("date").apply(wtret,include_groups=False)

aum_returns = pd.DataFrame(aum_ret).reindex(common_returns.index)
aum_rows = []
for cat in common_returns.columns:
    z = pd.concat([common_returns[cat],aum_returns[cat]],axis=1).dropna()
    z.columns = ["EW","AUM"]
    rff = factors.set_index("date")["RF"].reindex(z.index)
    ew = series_stats(z.EW,rff)
    aw = series_stats(z.AUM,rff)
    aum_rows.append([
        cat,z.EW.corr(z.AUM),ew[0],aw[0],ew[1],aw[1]
    ])

aum_robustness = pd.DataFrame(
    aum_rows,
    columns=["category","corr_EW_AUM","EW_return","AUM_return","EW_vol","AUM_vol"]
)
display(aum_robustness.round(4))
```
